## 1. Presentación del proyecto

Este proyecto desarrolla una prueba de concepto (POC) para explorar el uso de técnicas de generación de prompts aplicadas al proceso editorial.

La propuesta parte de una problemática vinculada a la publicación independiente de una obra literaria: la necesidad de realizar correcciones de texto, preservar la voz de la autora, verificar la coherencia de información biológica incorporada en los relatos y desarrollar una propuesta visual para la obra.

Para la implementación se utilizará como caso de estudio ficticio la obra *INSECTA – Voces de un mundo invisible*, perteneciente a una autora argentina nacida en Mendoza.

La POC se desarrollará sobre un cuento ficticio protagonizado por una libélula y permitirá demostrar la aplicación de modelos de generación de texto y de imagen mediante una secuencia de prompts.


### Nota sobre el caso de estudio

Para desarrollar la prueba de concepto no se dispone del manuscrito original de la obra. Por este motivo, se construyó un caso de estudio ficticio específicamente para evaluar la metodología propuesta.

El cuento y los materiales literarios utilizados en la POC son originales y ficticios. La información biológica utilizada como referencia para el proceso de verificación se basa en fuentes científicas.

Esta diferenciación permite evaluar el funcionamiento de los prompts sin atribuir a una autora real contenidos que no fueron proporcionados para este proyecto.


### 1.1 Cronograma de implementación

La prueba de concepto se desarrolla sobre un único cuento ficticio, seleccionado como caso representativo de la metodología propuesta.

La elección de un solo cuento permite concentrar el análisis en profundidad y demostrar el funcionamiento de la secuencia de prompts antes de extenderla a una obra completa.

| Etapa | Actividad | Duración estimada | Cantidad de cuentos | Iteraciones |
|---|---|---:|---:|---:|

| 2 | Corrección ortotipográfica | 1 jornada | 1 | 1 |
| 3 | Evaluación y revisión del resultado | 1 jornada | 1 | 1 |
| 4 | Verificación biológica y editorial | 1 jornada | 1 | 1 |

En total, la POC contempla un cuento, dos iteraciones principales de revisión textual y una generación de imagen, además de las instancias de evaluación y validación humana.

El cronograma corresponde a una estimación para la prueba de concepto y puede ampliarse en una implementación editorial real según la cantidad de cuentos y las necesidades de cada manuscrito.

### 1.2 Objetivos

### Objetivo general

Desarrollar una prueba de concepto que demuestre la aplicación de técnicas de generación de prompts para asistir distintas etapas de un proceso editorial independiente, combinando revisión textual, verificación de información biológica y generación de una propuesta visual.

### Objetivos específicos

- Diseñar prompts específicos para diferentes necesidades editoriales.
- Aplicar técnicas de asignación de roles y restricciones explícitas para orientar las respuestas del modelo.
- Preservar la voz narrativa y las decisiones estilísticas de la autora durante la revisión textual.
- Contrastar las afirmaciones biológicas presentes en el cuento con una ficha de referencia y fuentes externas.
- Incorporar una instancia de revisión humana para validar los resultados obtenidos mediante inteligencia artificial.
- Transformar elementos narrativos del cuento en una propuesta visual mediante un modelo generativo de imágenes.
- Documentar un flujo de trabajo reproducible que pueda utilizarse como base para futuros proyectos editoriales.

### 2.2. Texto original

El siguiente texto constituye el material literario de entrada de la POC. (Se trata de un cuento ficticio creado específicamente para este proyecto)

#### *El agua que recuerda*

La primera vez que vi la libélula pensé que había salido del agua.

Después comprendí que no era exactamente así.

Había llegado hasta el estanque aquella tarde de enero, cuando el calor de Mendoza parecía quedarse suspendido sobre los árboles. El agua estaba quieta y apenas se movía con el viento. Me senté en una piedra, como hacía tantas veces cuando necesitaba ordenar alguna idea, y entonces apareció ella.

Azul, fina y brillante.

Voló sobre la superficie del agua y se detuvo unos segundos en la punta de una rama. Sus alas transparentes parecían sostenerla en un lugar que no pertenecía ni al cielo ni a la tierra.

La observé durante un largo rato.

Pensé que había algo extraño en verla allí. Nosotros solemos imaginar que los insectos comienzan su vida cuando los vemos volar. Como si antes de tener alas no hubieran sido todavía quienes estaban destinados a ser.

Pero la libélula había conocido primero el agua.

Mucho antes de aquel vuelo había sido pequeña, silenciosa y acuática. Había vivido bajo la superficie, entre plantas y sombras, cazando pequeños animales que encontraba en aquel mundo que para nosotros permanece escondido.

Durante meses, quizás durante años, el estanque había sido su universo.

Yo miraba el agua e intentaba imaginarla allí abajo.

Una criatura sin alas.

Una criatura que todavía no sabía que algún día tendría el cielo entero por delante.

La libélula volvió a moverse.

Esta vez no voló en línea recta. Dibujó una curva sobre el estanque, desapareció detrás de un grupo de juncos y volvió a aparecer unos metros más lejos. Parecía jugar con el aire, aunque seguramente solo estaba haciendo lo que su naturaleza le había enseñado: buscar, perseguir, alimentarse, sobrevivir.

Me gustó pensar que incluso aquello que nosotros llamamos belleza tiene una explicación.

Y que tener una explicación no la hace menos hermosa.

El sol comenzó a bajar.

Entonces miré nuevamente el agua.

Había pasado tantos años creyendo que las cosas importantes sucedían cuando finalmente podían verse, que me sorprendió descubrir lo contrario. La mayor parte de la vida de aquella criatura había ocurrido lejos de mis ojos.

Debajo de la superficie.

En silencio.

Esperando.

Quizás por eso aquella libélula me conmovió.

Porque había pasado gran parte de su existencia en un lugar que nadie miraba y, sin embargo, cuando finalmente salió al mundo, parecía llevar consigo todo lo que había vivido.

Se posó nuevamente sobre la rama.

Yo no sabía si sería la misma libélula que había visto unos minutos antes. Tampoco sabía cuánto tiempo viviría después de aquel momento.

Pero entendí algo.

El agua no era solamente el lugar del que había salido.

Era también el lugar que la había formado.

Me quedé allí hasta que comenzó a oscurecer.

Cuando finalmente me levanté, la libélula ya no estaba.

El estanque volvió a quedar quieto.

Y durante unos segundos tuve la sensación absurda de que el agua guardaba memoria.

Quizás todas las cosas importantes funcionan de esa manera.

Primero ocurren en silencio.

Después aparecen.

Y cuando finalmente podemos verlas, ya llevan mucho tiempo sucediendo.

## 3. Módulo Texto-Texto

### 3.1. Prompt 1 — Corrección ortotipográfica

El primer paso de la POC consiste en realizar una revisión ortotipográfica del cuento.

El objetivo es detectar errores de ortografía, gramática, sintaxis y puntuación, procurando conservar la estructura narrativa, el tono y las elecciones estilísticas de la autora.

Para esta etapa se utiliza una estrategia de asignación de rol y restricciones explícitas, indicándole al modelo qué tipo de modificaciones puede realizar y cuáles debe evitar.

### Prompt utilizado

> Actúa como un editor ortotipográfico profesional especializado en literatura en español.
>
> Tu tarea es revisar el cuento proporcionado y detectar únicamente errores ortográficos, gramaticales, sintácticos y de puntuación.
>
> Es fundamental preservar la voz narrativa de la autora. No reescribas el cuento, no modifiques su tono literario, no sustituyas expresiones por preferencias estilísticas propias y no alteres la estructura narrativa.
>
> Cuando una expresión sea válida desde el punto de vista lingüístico, aunque pueda formularse de otra manera, debes conservarla.
>
> Realiza las correcciones estrictamente necesarias.
>
> Al finalizar:
> 1. presenta el cuento corregido;
> 2. incluye una lista breve de los cambios realizados;
> 3. explica cuando no haya sido necesario modificar un fragmento por tratarse de una elección estilística válida.

### Resultado — Primera iteración

A continuación se presenta el resultado obtenido al aplicar el Prompt 1 sobre la Versión 0 del cuento.

El resultado será posteriormente comparado con el texto original para identificar qué modificaciones realizó el modelo y determinar si respetó las restricciones establecidas en el prompt.
### Resultado obtenido

El modelo determinó que el cuento no presentaba errores ortográficos, gramaticales ni sintácticos que requirieran corrección. Por este motivo, conservó íntegramente el texto original.

**Resultado de la revisión:**

* Ortografía: sin errores detectados.
* Gramática: sin errores detectados.
* Sintaxis: construcciones correctas.
* Puntuación: adecuada al texto y a su ritmo narrativo.
* Estructura: conservada íntegramente.

El modelo también identificó fragmentos narrativos que constituyen elecciones estilísticas válidas, como las oraciones breves, las repeticiones y el uso de conjunciones al inicio de algunos párrafos. Por lo tanto, no realizó modificaciones que pudieran alterar la voz narrativa de la autora.

**Conclusión de la primera iteración:** el modelo cumplió con las restricciones principales del prompt, ya que mantuvo el texto original y evitó realizar cambios estilísticos innecesarios.


### 3.2. Evaluación de la primera iteración

### 3.3. Prompt 2 - Verificación biológica
El segundo paso de la POC consiste en analizar la coherencia narrativa del cuento y contrastar las afirmaciones biológicas con una ficha de referencia.

El objetivo es identificar posibles contradicciones entre el relato y la información biológica, sin alterar la voz narrativa ni modificar innecesariamente el contenido literario.

Para esta etapa se utiliza una estrategia de asignación de roles, delimitación de información y restricciones explícitas. El modelo deberá diferenciar los aspectos científicos de las decisiones propias de la ficción.


### Prompt utilizado

Actúa como editor literario especializado en divulgación científica y como asistente de revisión de contenidos biológicos.

Tu tarea es analizar el cuento proporcionado y contrastar sus afirmaciones sobre las libélulas con la ficha biológica de referencia.

Trabaja con las siguientes instrucciones:

1. **Verificación biológica:** identifica las afirmaciones relacionadas con la biología de las libélulas y contrástalas con la ficha proporcionada. No inventes datos ni des por ciertas afirmaciones que no puedan verificarse con la información disponible.

2. **Coherencia narrativa:** analiza si la información biológica está integrada de forma comprensible en el relato y si existen contradicciones entre sus distintas partes.

3. **Preservación literaria:** respeta la voz narrativa, el tono y la estructura del cuento. No reescribas el texto ni sustituyas expresiones por preferencias estilísticas.

4. **Identificación de observaciones:** clasifica los hallazgos en tres categorías:

   * Información coherente con la ficha biológica.
   * Información que requiere verificación adicional.
   * Posibles inconsistencias biológicas.

5. **Propuestas de mejora:** si detectas alguna inconsistencia, indica el fragmento correspondiente y explica el motivo. Propón una modificación únicamente cuando sea necesaria, sin alterar el estilo literario.

Presenta el resultado en cuatro apartados:

* Afirmaciones biológicas identificadas.
* Análisis de coherencia y posibles inconsistencias.
* Observaciones y propuestas de mejora.
* Conclusión editorial.

Si la ficha no contiene información suficiente para confirmar una afirmación, indícalo expresamente. No confundas una licencia literaria con un error científico.



### Resultado — Segunda iteración

El modelo analizó el cuento *El agua que recuerda* y contrastó sus afirmaciones biológicas con la ficha de referencia.

**Principales resultados:**

* **Alimentación:** las referencias a la depredación durante la etapa juvenil se consideraron compatibles con la información disponible.

* **Longevidad adulta:** no puede determinarse con la ficha proporcionada.

**Conclusión del modelo:**

El modelo no identificó contradicciones biológicas demostrables que exigieran modificar el cuento. Recomendó conservar la narración y verificar las afirmaciones que exceden la información disponible.

## 5. Metodología

La metodología se organizó como un flujo secuencial de trabajo compuesto por tres módulos principales: corrección de texto, verificación de contenido y generación de imagen.

### 5.1. Preparación de los datos

Se construyó un caso de estudio ficticio compuesto por:

- un cuento literario titulado *El agua que recuerda*;
- una ficha biológica de referencia sobre libélulas;
- información contextual ficticia sobre la obra *INSECTA – Voces de un mundo invisible*.

El cuento y la información relacionada con la autora forman parte del caso ficticio desarrollado específicamente para la POC. La información biológica fue diferenciada de estos elementos y posteriormente contrastada con fuentes externas.

### 5.2. Aplicación de técnicas de generación de prompts

Los prompts fueron diseñados incorporando diferentes técnicas de estructuración:

- **Asignación de rol:** se define el perfil que debe adoptar el modelo, por ejemplo, editor ortotipográfico o asistente de revisión biológica.
- **Restricciones explícitas:** se especifican las acciones que el modelo debe realizar y aquellas que debe evitar.
- **Contexto delimitado:** se separan claramente el cuento y la ficha biológica para facilitar el análisis.
- **Salida estructurada:** se indican los apartados y categorías que debe presentar el modelo.
- **Iteración y evaluación:** los resultados obtenidos se analizan antes de avanzar a la siguiente etapa.

Estas técnicas permiten reducir respuestas ambiguas y orientar el modelo hacia el objetivo específico de cada tarea.

### 5.3. Revisión humana

La inteligencia artificial se utilizó como herramienta de asistencia y no como instancia definitiva de decisión editorial.

En la revisión ortotipográfica se comprobó que el modelo no modificara innecesariamente la voz narrativa.
